# <center>What is Python ?

Python is a high-level, interpreted programming language known for its **simplicity and readability**. It was created by Guido van Rossum and first released in 1991. Python emphasizes code readability and allows developers to express concepts in fewer lines of code compared to languages like C++ or Java.

## Key Features:
- (1) **Interpreted Language**: Python code is executed line by line, which makes debugging easier
- (2) **Versatile**: Python supports **multiple programming paradigms**, including procedural, object-oriented, and functional programming
- (3) **Dynamically Typed**: You don’t need to **declare variable types explicitly**; Python infers them at runtime
- (4) **Extensive Libraries**: Python has a rich standard library and a vast ecosystem of third-party libraries (e.g., **NumPy**, Pandas, TensorFlow, Django) for various tasks

## (7) Classes

- A **class** is a blueprint for creating objects. It defines attributes (data) and methods (functions) that describe the behavior of an object.
- use `class` keyword to define a class

### Key concepts:
1. Class → A blueprint for objects.
2. Object → An instance of a class.
3. Attributes (Variables) → Store object data (`self.brand`).
4. Methods (Functions in Class) → Define object behavior (`display_info()`).
5. Constructor (`__init__`) → Initializes object attributes.

**Note**: you may need to use `self` to let the interpreter understands that the **attributes** and **methods** will belong to each created object. without it it will be considered as **static** or **Class variable** which is shared with all instances !

### Encapsulation

- Use private attributes (`__attribute`) to hide data.

```python
class Dog:
    species = "Mammal"  # Class Variable

    def __init__(self, name, age):
        self.name = name  # Instance Variable
        self.__age = age  # Private attribute
        

dog1 = Dog("Buddy", 2)
dog2 = Dog("Max", 1)

print(dog1.name)  # Output: Buddy
print(dog2.name)  # Output: Max
print(dog1.species, dog2.species) # Output: Mammal Mammal
Dog.species = "German"
print(dog1.species, dog2.species)
print(dog1.age)
```

```python
# Inheritance
class Animal:
    def __init__(self, species):
        self.species = species
        print(f"Animal of species {self.species} is created.")

class Dog(Animal):
    def __init__(self, species, breed):
        # Call the parent class constructor
        super().__init__(species)            # super() keyword in order to refer to parent class constructor
        self.breed = breed
        print(f"Dog of breed {self.breed} is created.")

# Creating an object of Dog
dog = Dog("Canine", "Labrador")

# Output:
# Animal of species Canine is created.
# Dog of breed Labrador is created.
```

```python
# Multiple-Inheritance
class Parent1:
    def __init__(self):
        self.var = "Benha University"

class Parent2:
    def __init__(self):
        self.var = "FCAI"

class Child(Parent1, Parent2): #The order of inheritance is CRITICAL to avoid Diamond Problem (Method Resolution Order -MRO-)
    def __init__(self):
        super().__init__()
        print(self.var) # Whose (var) will be printed ????????

x = Child()
```

```python
# Another example but with reversed MRO
class Child(Parent2, Parent1): #Reversed order
    def __init__(self):
        super().__init__()
        print(self.var) # Whose (var) will be printed ????????

x = Child()
```

* If incosistency found during Multiple-Inheritance, "MRO" allow for **((First Parent Only))** to pass through its copy
* It is a method **"MRO"** instead of `Virtual Inheritance`  in C++

```python
#Polymorphism
class Vehicle:
    def __init__(self, brand, model):
        self.model = brand
        self.brand = model

    def move(self): # No need to define a pure virtual function (just like C++) -Dynamicity Feature-
        pass

    def info(self):
        print(self.model, self.brand)

class Car(Vehicle):
    def __init__(self, brand, model):
        super().__init__(brand,model)
        
    def move(self): #Function overriding
        print("Move !")

class Boat(Vehicle):
    def __init__(self, brand, model):
        super().__init__(brand,model)
        
    def move(self): #Function overriding
        print("Sail !")

class Plane(Vehicle):
    def __init__(self, brand, model):
        super().__init__(brand,model)
        
    def move(self): #Function overriding
        print("Fly !")

# No need to change info() method since it is common
```

```python
car1 = Car("Ford", "Mustang")
boat1 = Boat("Ibiza", "Touring 20")
plane1 = Plane("Boeing", "747")
for x in (car1, boat1, plane1): # no need to define (x) as reference to parent class -Dynamicity Feature-
    x.info()
    x.move()
```
